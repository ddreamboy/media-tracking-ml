```python
import pandas as pd
```

```python
data_path = "results/inference_demo/posts_2026_labeled_64958.parquet"
df_2026_labeled = pd.read_parquet(data_path)
```

```python
for row in df_2026_labeled.head(100).itertuples():
    print(f"Text: {row.text}")
    print(f"Label: {row.label}")
    print(f"Zone: {row.zone}")
    print(f"Confidence: {row.confidence}")
    print("-" * 50)
```

```python
df_2026_labeled["label"] = df_2026_labeled["label"].apply(
    lambda x: (
        x.replace(" [периферия гетерогенного кластера]", "")
        if isinstance(x, str)
        else x
    )
)
```

```python
for row in df_2026_labeled.head(100).itertuples():
    print(f"Text: {row.text}")
    print(f"Label: {row.label}")
    print(f"Zone: {row.zone}")
    print(f"Confidence: {row.confidence}")
    print("-" * 50)


df_2026_labeled.to_parquet(
    "results/inference_demo/posts_2026_labeled_cleaned.parquet", index=False
)
```
