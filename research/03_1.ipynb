```python
from sentence_transformers import SentenceTransformer
import pandas as pd
import numpy as np
```

```python
df = pd.read_parquet("data/posts_preprocessed.parquet")
```

```python
df.info()
```

```python
model = SentenceTransformer("deepvk/USER-bge-m3")

docs_for_embed = df["text_clean"].tolist()

embeddings = model.encode(
    docs_for_embed, batch_size=64, show_progress_bar=True, normalize_embeddings=True
)
```

```python
np.save("data/embeddings_USER-bge-m3_300k.npy", embeddings)
print(f"Готово: {embeddings.shape}")
```

```python
df = df.reset_index(drop=True)
df["embed_idx"] = df.index

df[["embed_idx", "id_post"]].to_parquet("data/embed_index_map.parquet", index=False)
print("Маппинг сохранён")
```
